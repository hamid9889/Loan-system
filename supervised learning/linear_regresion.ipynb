```python
import numpy as np
```

```python
#linerar regression model

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.weights = None
        self.bias = None
        self.learning_rate = learning_rate
        self.n_iter = n_iter
        
    def fit(self, X, y): #x _train , Y_trsin
        n, n = X.shape
         # step 1 initilize
        self.bias = 0
        self.weights = np.zeros(n) #[0,0,0]
        
        # gradient decent
        for i in range(self.n_iter):
            # calc y pred 
            y_pred = self.bias + np.dot(X, self.weights)
             
             #  step -conversion theorem 
            db = (1/n) * np.sum(y_pred - y)
            dw = (1/n) * np.dot(X.T, (y_pred - y))
        
        
        
        self.bias -= self.learning_rate * db
        self.weights -= self.learning_rate * dw
        
       
        
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4],[5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(f"pridicted value:{y_pred}")

print(f"End weights: {model.weights}")
print(f"End bias: {model.bias}")
```

```python
#linesr regression with ols

class LinearRegressionOLS:
    def __init__(self):
        self.weights = None
        self.bias = None
        
    def fit(self, X, y):
        m,n = X.shape
        
        X_b = np.c_[np.ones((m, 1)), X] # add bias term
        
        
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
         
        self.bias = theta[0]
        self.weights = theta[1:]    
             
             
             
             
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred             
```

```python
X = np.array([[1], [2], [3], [4],[5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print(f"pridicted value:{y_pred}")

print(f"End weights: {model.weights}")
print(f"End bias: {model.bias}")
```
